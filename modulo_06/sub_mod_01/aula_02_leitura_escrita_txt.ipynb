Fora do escopo do mundo/contexto python, dependendo de onde é  executa determinado programa ou 
processado determinado arquivo, este pode ser afetado quanto a qual carácter de "Nova Linha" será efetivamente gravado ou lido.

Em sistemas operacionais Windows se utiliza como padrão o carácter:
[CRLF] : Que é a quebra de linha e retorno do carro ("posição  do  cursor" no arquivo

Em sistemas operacionais Linux  se utiliza como padrão o carácter:
[LF] : Que é a quebra de linha

Sendo assim,  neste momento de aprendizagem irei realizar o uso do módulo **os** do Python,  para obter o carácter de quebra de linha do sistema operacional no qual este conteúdo estiver sendo executado:

```python
from os import linesep

# CONSTANTES
DELIMITADOR_ESCOPOS_IO_CONSOLE = str("*" * 50)
DELIMITADOR_ESCOPOS_IO_ESCRITA_CONTEUDO = str("|" * 50)


# VARIAVEIS

# Linux: A partir do do diretorio atual,  se sobe um nível, chegando-se a pasta destino/origem
caminho_origem = "../repositorio"
# nome do(s) arquivo(s); Passar nome e extensão
arquivos = [
    "aula_02_leitura_escrita_from_ipynb.txt",
    "aula_02_leitura_escrita_from_ipynb.csv",
]
```

No bloco abaixo, replico o conhecimento observado em aula, onde é demonstrado o uso da função *open()*.
No uso da função e para o parâmetro __mode__, estou utilizando a opção "w", que tem como definição a sobre escrita de todo o conteúdo do arquivo.

**__Nota__**: 
1 - Se o arquivo não existe, a função vai criar o arquivo e seguir com execução de escrita
     Se o arquivo existir, todo o conteúdo dele, sera apagado, e a execução de escrita vai continuar

```python
# Definicao
def escrevendo_arquivo_txt(caminho_origem, arquivos):
    print("Iniciando processamento | escrevendo_arquivo_txt()")

    for arquivo in arquivos:
        print(f"\tProcessando:\t{arquivo}")

        arquivo_em_processamento = f"{caminho_origem}/{arquivo}"
        with open(arquivo_em_processamento, "w") as destino:
            destino.write(f"Escrevendo um linha em '{arquivo}'")

        print(f"\t  Processado:\t{arquivo}")
    print("Terminado processamento | escrevendo_arquivo_txt()")


# Uso
escrevendo_arquivo_txt(caminho_origem, arquivos)
```

No bloco abaixo, novamente replico o conhecimento observado em aula, onde é demonstrado o uso da função *open()*.
Mas neste bloco para  o parâmetro __mode__, estou utilizando a opção "a", que tem como definição a  escrita de inclusao de  conteúdo no(s) arquivo(s).

**__Nota__**: 
1 -   Comportamento padrão da opção "a" é apenas adicionar o conteúdo ao final do arquivo. Caso seja necessario criar uma nova linha, a string/valor 
deve especificar o caracter de quebra de linha.

```python
# Definicao
def adicionando_conteudo_arquivo_txt(caminho_origem, arquivos):
    print("Iniciando processamento | adicionando_conteudo_arquivo_txt()")
    caracter_quebra_linha = linesep

    for arquivo in arquivos:
        print(f"\tProcessando: {arquivo}")

        arquivo_em_processamento = f"{caminho_origem}/{arquivo}"
        with open(arquivo_em_processamento, "a") as destino:
            destino.write(
                f"{caracter_quebra_linha}Adicionando novo conteudo ao arquivo '{arquivo}'"
            )
        print(f"\tProcessado: {arquivo}")
    print("Terminado processamento | adicionando_conteudo_arquivo_txt()")

    print(f"\n{DELIMITADOR_ESCOPOS_IO_CONSOLE}")


# Uso
adicionando_conteudo_arquivo_txt(caminho_origem, arquivos)
```

No bloco abaixo, é utilizado a função open(), porém, com parametro mode como "r", o que resulta em acesso de somente leitura do arquivo.
Conforme nome da função criada "lendo_todo_conteudo_arquivo_txt", é que todo o conteúdo é carregado em memoria de uma unica vez.  Esta leitura é realizada pelo uso da funcao read() do objeto TextIOWrapper.

```python
# Definicao
def lendo_todo_conteudo_arquivo_txt(caminho_origem, arquivos):
    print("Iniciando  processamento | lendo_todo_conteudo_arquivo_txt()")
    for arquivo in arquivos:
        arquivo_em_processamento = f"{caminho_origem}/{arquivo}"
        print(f"\n\tLendo arquivo: '{arquivo}'")
        print(f"\tConteudo:")
        with open(arquivo_em_processamento, "r") as origem:
            conteudo = origem.read()
            print(DELIMITADOR_ESCOPOS_IO_ESCRITA_CONTEUDO)
            print(f"{conteudo}'")
            print(DELIMITADOR_ESCOPOS_IO_ESCRITA_CONTEUDO)
        print(f"\tLido arquivo: '{arquivo}'")
    print("Terminado processamento | lendo_todo_conteudo_arquivo_txt()")


# Uso
print(f"\n{DELIMITADOR_ESCOPOS_IO_CONSOLE}")
lendo_todo_conteudo_arquivo_txt(caminho_origem, arquivos)
```

No bloco abaixo, ainda é utilizado a função open(), tambem com parametro mode como "r", o que resulta em acesso de somente leitura do arquivo.
Mas conforme nome da função criada "lendo_linha_a_linha_conteudo_arquivo_txt", o conteúdo é carregado em memoria linha a linha atraves do uso da funcao readlines() do objeto TextIOWrapper.

```python
# Definicao
def lendo_linha_a_linha_conteudo_arquivo_txt(caminho_origem, arquivos):
    print("Iniciando  processamento | lendo_linha_a_linha_conteudo_arquivo_txt()")
    for arquivo in arquivos:
        arquivo_em_processamento = f"{caminho_origem}/{arquivo}"
        print(f"\n\tLendo arquivo: '{arquivo}'")
        print(f"\tConteudo:")
        with open(arquivo_em_processamento, "r") as origem:
            print(DELIMITADOR_ESCOPOS_IO_ESCRITA_CONTEUDO)
            for index, linha in enumerate(origem.readlines()):
                conteudo = linha.strip()
                print(f"Linha: {index} | {conteudo}")
            print(DELIMITADOR_ESCOPOS_IO_ESCRITA_CONTEUDO)
        print(f"\tLido arquivo: '{arquivo}'")
    print("Terminado processamento | lendo_linha_a_linha_conteudo_arquivo_txt()")


# Uso
print(f"\n{DELIMITADOR_ESCOPOS_IO_CONSOLE}")
lendo_linha_a_linha_conteudo_arquivo_txt(caminho_origem, arquivos)
```
